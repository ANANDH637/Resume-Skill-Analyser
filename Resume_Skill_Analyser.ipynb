```python
import spacy
import re
from collections import Counter

nlp = spacy.load("en_core_web_sm")

print("Resume Skill Analyzer Loaded Successfully!")
```

```python
skill_categories = {
    "Programming": [
        "python", "java", "c", "c++", "javascript", "sql",
        "r", "html", "css"
    ],

    "Data & AI": [
        "machine learning", "deep learning", "artificial intelligence",
        "data science", "nlp", "tensorflow", "pytorch",
        "pandas", "numpy", "scikit-learn"
    ],

    "Tools & Platforms": [
        "git", "github", "docker", "linux", "aws", "azure",
        "google cloud", "jupyter", "google colab"
    ],

    "Databases": [
        "mysql", "mongodb", "postgresql", "oracle",
        "sqlite", "database"
    ],

    "Web & Frameworks": [
        "react", "node.js", "flask", "django",
        "bootstrap", "rest api"
    ]
}

print("Skill database created!")
```

```python
resume_text = """
I am a Computer Science student with strong knowledge in Python,
C++, SQL and Java. I have worked on machine learning and natural
language processing projects.

I have experience with Pandas, NumPy, Scikit-learn and TensorFlow.
I use Git and GitHub for version control and Google Colab for
machine learning experiments.

I have also worked with MySQL, MongoDB and Flask. My projects
include a chatbot, student performance prediction system and
text classification application.
"""

print("Resume loaded successfully!")
```

```python
resume_lower = resume_text.lower()

detected_skills = {}

for category, skills in skill_categories.items():
    found = []

    for skill in skills:
        if skill.lower() in resume_lower:
            found.append(skill)

    detected_skills[category] = found

print("Detected Skills:\n")

for category, skills in detected_skills.items():
    print(f"{category}:")

    if skills:
        for skill in skills:
            print("  ✓", skill)
    else:
        print("  No skills detected")

    print()
```

```python
total_skills = sum(len(skills) for skills in skill_categories.values())
detected_count = sum(len(skills) for skills in detected_skills.values())

skill_score = (detected_count / total_skills) * 100

print("========== RESUME SKILL SCORE ==========")
print(f"Skills Detected : {detected_count}")
print(f"Skills Database : {total_skills}")
print(f"Skill Coverage  : {skill_score:.2f}%")
```

```python
print("========== SUGGESTED SKILLS ==========\n")

for category, skills in skill_categories.items():

    missing = [
        skill for skill in skills
        if skill not in detected_skills[category]
    ]

    if missing:
        print(f"{category}:")
        print("  →", ", ".join(missing))
        print()
```

```python
print("========== CATEGORY ANALYSIS ==========\n")

for category, skills in detected_skills.items():

    total = len(skill_categories[category])
    found = len(skills)

    percentage = (found / total) * 100

    print(f"{category}")
    print(f"Skills Found : {found}/{total}")
    print(f"Coverage     : {percentage:.1f}%")
    print("-" * 35)
```

```python
if skill_score >= 70:
    rating = "Excellent"
elif skill_score >= 50:
    rating = "Good"
elif skill_score >= 30:
    rating = "Average"
else:
    rating = "Needs Improvement"

print("""
╔══════════════════════════════════════╗
        RESUME SKILL ANALYZER
╚══════════════════════════════════════╝
""")

print(f"Total Skills Detected : {detected_count}")
print(f"Skill Coverage        : {skill_score:.2f}%")
print(f"Overall Rating        : {rating}")

print("\nDetected Skills:")

for category, skills in detected_skills.items():
    if skills:
        print(f"\n{category}")
        print("  " + ", ".join(skills))

print("\n======================================")
print("Resume analysis completed successfully!")
```
